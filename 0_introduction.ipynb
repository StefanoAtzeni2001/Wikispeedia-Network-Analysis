# Wikispeedia Network Analysis: Introduction 

## Wikipedia 
Wikipedia is a collaborative online encyclopedia that allows users to create and consult articles on a wide range of topics. Beyond sharing human knowledge in an open and accessible manner, it is often used as a source of semantic data, not only as a large textual corpus, but also for its network structure. Individual articles can be considered as nodes in a network, and the hypertext links between articles as edges connecting them. The semantic network structure of Wikipedia is highly interesting and has been the subject of study in numerous research papers.

## Wikispeedia Dataset
The wiki game is a simple online game where players start from one Wikipedia page and must reach another page by clicking on links present in the pages. One of the most famous websites for playing this game is [Wikispeedia](https://wikispeedia.org/), from which the [Wikispeedia navigation paths](https://snap.stanford.edu/data/wikispeedia.html) dataset from Stanford University was extracted.

The dataset contains a reduced version of Wikipedia from 2007 and the completed and uncompleted paths taken by Wikispeedia users.

- Finished paths: 51,318
- Unfinished paths: 24,875
- Articles: 4,604
- Links: 119,882

## Objective
The objective of this project is to analyze the network structure of Wikipedia in the context of Wikispeedia. The analysis will be exploratory in nature, seeking to obtain insights into the network structure and how users navigate within it.
Some key questions that we will attempt to answer are:

- **1_graph_properties.ipynb**: What type of structure does the Wikipedia network have?
- **2_nodes_centralities.ipynb**: Which articles are most central in the network?
- **3_community_detection.ipynb**: Are Wikipedia categories reflected in a community structure?
- **4_paths_difficulty.ipynb**: The difficulty of a challenge is correlated with structural or semantic difficulty?
- **5_hubs_in_paths.ipynb**: The nodes most frequently visited by players are also the most central?

## Future Developments
There are infinite possible developments in exploring the Wikipedia network. Some potential directions could be:

- **Using an updated dataset**: The dataset used is from a 2007 version of Wikipedia, almost 20 years ago. The number of articles is extremely reduced compared to the current one (approximately 7 million), which made the network more manageable, but the main problem lies in poorly developed articles. Initially, there were no standard rules for contributing to articles, so many articles are very brief and contain few arbitrary links, affecting the quality of the network.

- **Creating an interactive visualization**: The Wikipedia network is very large and complex, and an interactive visualization could help explore the network more intuitively. Some attempts were made using d3.js, but due to the size of the network, the visualization froze and became unusable. Due to time constraints, this attempt was not pursued further.

- **Using other community detection algorithms**: The Louvain algorithm was used as it is more commonly used and faster, but other algorithms such as Clauset-Newman-Moore or Label Propagation could be tested. Additionally, different Wikipedia subcategories could be used.

- **Integration of semantic content**: NLP techniques could be used to analyze the textual content of articles instead of hypertext links, to connect articles or weight the network edges.

- **Strategy analysis**: User navigation paths could be analyzed more deeply, for example, trying to understand the strategies used and searching for a possible optimal strategy to reach a page.
